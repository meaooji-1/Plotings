```python
import pandas as pd
df=pd.DataFrame({
    
    "Patient":["p1","p2","p3","p4","p5","p6","p7","p8"],
    "Age":[25,35,45,30,50,40,28,55],
    "Glucose":[95,110,180,98,210,150,105,250],
    "BMI":[22.5,24.8,30.2,23.0,32.1,28.5,21.8,34.6],
    "BP":[120,125,145,118,155,140,122,170],
    "Gender":["male","Female","male","Female","male","Female","Female","male"],
    "Disease":["yes","no","yes","no","yes","yes","no","yes"]
})
print(df)
```

```python
print(df.ndim)
print(df.shape)
print(df.size)
```

```python
import matplotlib.pyplot as plt
plt.plot(df["Patient"],df["Glucose"],marker="o")
plt.title("Patient_Glucose Level")
plt.xlabel("patient")
plt.ylabel("Glucose level")
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.bar(df["Patient"],df["Glucose"],color="red")
plt.title("Patient_Glucose Level")
plt.xlabel("patient")
plt.ylabel("Glucose level")
plt.grid(False)
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.barh(df["Patient"],df["Glucose"],color="red")
plt.title("Patient_Glucose Level")
plt.xlabel("patient")
plt.ylabel("Glucose level")
plt.grid(False)
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.scatter(df["Patient"],df["Glucose"],color="red")
plt.title("Patient_Glucose Level")
plt.xlabel("patient")
plt.ylabel("Glucose level")
plt.grid(False)
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.hist(df["Glucose"],bins=5,edgecolor="red")
plt.title("Glucose_Distribution Level")
plt.ylabel("Glucose level")
plt.grid(False)
plt.show()
```

```python
import matplotlib.pyplot as plt
df["Glucose"].value_counts().plot(kind="pie",autopct="%1.1f%%")
plt.title("Glucose")
plt.ylabel=("")
plt.show()
```

```python
plt.boxplot(df["Glucose"])
plt.title("Glucose Boxplot")
plt.show()
```

```python
import seaborn as sns
sns.violinplot(y=df["Glucose"])
plt.title("Violin Plot")
plt.show()
```

```python
#categorical data
sns.countplot(x=df["Disease"])
plt.title("Count of yes and no Plot")
plt.show()
```

```python
sns.countplot(x=df["Gender"])
plt.title("Count of Gender")
plt.show()
```

```python
a=df.corr(numeric_only=True)

sns.heatmap(
    a,
    annot=True,
    cmap="coolwarm"
)
plt.show()
```
