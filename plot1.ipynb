```python
import matplotlib.pyplot as plt
x = [1, 2, 3, 4]
y = [10, 15, 7, 10]
plt.plot(x, y, marker='o')
plt.title("Line Plot")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.grid(True)
plt.show()
```
