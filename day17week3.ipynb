```python
import pandas as pd
df= pd.read_csv("employee_coding_problems_dataset.csv")
print(df)
```

```python
Q1=df["Salary"].quantile(0.25)
print(Q1)
Q2=df["Salary"].quantile(0.50)
print(Q2)
Q3=df["Salary"].quantile(0.75)
print(Q3)
```

```python
Q1=df["Age"].quantile(0.25)
print(Q1)
Q2=df["Age"].quantile(0.50)
print(Q2)
Q3=df["Age"].quantile(0.75)
print(Q3)
```

```python
IQR=Q3 -Q1
print(IQR)
```

```python
lower = Q1 - 1.5 * IQR
upper = Q3 - 1.5 * IQR
print(lower,upper)
```

```python
df_no_outlier=df[(df["Age"]>=lower) & (df["Age"]<=upper)]
df_no_outlier




```
